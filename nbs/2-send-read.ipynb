```python
import requests
path="/api/events/"
baseurl="http://localhost:8002"
endpoint=f"{baseurl}{path}"
endpoint
```

```python
data={"page":"/test+"}
import json
json.dumps(data)
```

```python
response = requests.post(endpoint,data=json.dumps(data))
print(response.headers)
print(response.ok)
if response.ok:
    data= response.json()
    print(data, type(data), data.get('items'))
else:
    print(response.text)
```
