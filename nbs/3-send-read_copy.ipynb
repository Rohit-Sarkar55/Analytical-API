```python
import requests
```

```python
path="/api/events/"
baseurl="http://localhost:8002"
create_endpoint=f"{baseurl}{path}"
response = requests.post(create_endpoint, json={"page":"/tests"})
print(response.headers)
print(response.ok)
if response.ok:
    data= response.json()
    print(data)
else:
    print(response.text)
```

```python
detail_path="/api/events/3"
detail_endpoint=f"{baseurl}{detail_path}"
headers={
    "Content-Type":"application/json"
}
response = requests.put(detail_endpoint,json={"description":"testing update"},headers=headers) #putting headers is optional here unless you want to specify
print(response.headers)
print(response.ok)
if response.ok:
    data= response.json()
    print(data, type(data), data.get('items'))
else:
    print(response.text)
```

```python
import random

pages = ['/about','/home','/contact','pricing']

N = 100
for i in range(N):
    page = random.choice(pages)
    resp = requests.post(create_endpoint, json={"page":page})
    if resp.ok:
        data= response.json()
        print(data)
    else:
        print(response.text)
```
