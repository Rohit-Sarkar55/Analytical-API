```python
import requests
path="/api/events/"
baseurl="http://localhost:8002"
endpoint=f"{baseurl}{path}"
endpoint
```

```python
response = requests.get(endpoint)
print(response.text)
print(response.ok)
if response.ok:
    data= response.json()
    print(data, type(data), data.get('items'))
```
